```python
fig = plt.figure() #큰 도화지를 만듦

# 2: 도화지를 2개의 행으로 나눈다.
# 1: 한 행에 1개의 열이 있다.
# 1: 그 중에서 첫번째 칸을 사용한다.

fg1 = fig.add_subplot(2,1,1)
fg2 = fig.add_subplot(2,1,2)

fg1.plot([1,2,3],[4,5,6])
fg2.plot([1,2,3],[4,5,7])
```

```python
# 문제 2
# 1. 하나의 도화지를 생성한다.
# 2. 도화지를 1행 2열로 나눈뒤, 첫번째 칸에 그래프 영역을 하나 생성
# 3. 같은 도화지를 다시 2행 2열로 나눈뒤,
#    첫 번째 행의 두번째 칸에 그래프 영역을 하나 더 생성한다.

import matplotlib.pyplot as plt

fig = plt.figure() # 1. 하나의 도화지 생성

# 2. 도화지를 1행 2열로 나눈 뒤, 첫 번째 칸(1번)에 생성
fg1 = fig.add_subplot(1, 2, 1)

# 3. 도화지를 2행 2열로 나눈 뒤, 첫 번째 행의 두 번째 칸(2번)에 생성

fg2 = fig.add_subplot(2, 2, 2)

x = range(10)
y = range(5,15)

fg2.plot(x,y,"ch")
fg1.plot(x,y,"rp")
plt.show()
```

```python
fig = plt.figure()

plt.style.use('ggplot')

a = fig.add_subplot(111)

x = [3,4] #막대가 서 있는 위치(가로)
y = [1,2] #각 막대의 높이(세로)

labels = ['english','math']
colors = ['#9eaeff','#fff78a']

a.bar(x,y,tick_label=labels,color=colors)
plt.show()
```

```python
# 문제 3
# 막대 그래프는 가로 방향으로
# 막대의 값은 1,2이고 위치는 3,4
# 각 막대의 값은 각각 원하는 색으로 지정
# 라벨은 영어,수학으로 지정

import matplotlib.pyplot as plt

fig = plt.figure()

# 스타일 설정
plt.style.use('ggplot')

a = fig.add_subplot(111)

# 1. 변수 설정
# barh에서는 y가 위치, width(x)가 막대의 값입니다.
y = [3, 4]           # 막대가 놓일 위치 (문제 조건: 3, 4)
values = [1, 2]      # 막대의 값/길이 (문제 조건: 1, 2)

labels = ['english', 'math']
colors = ['#9eaeff', '#fff78a'] # 각각 지정한 색상

# 2. 가로 막대 그래프 그리기
# barh(위치, 값, 라벨, 색상) 순서로 적용합니다.
a.barh(y, values, tick_label=labels, color=colors)

for i,v in enumerate(y):
    a.text(v+2,i+1.05,str(v))
plt.show()
```
