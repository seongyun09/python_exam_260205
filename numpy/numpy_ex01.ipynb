```python
import numpy as np

arr = np.array([[3,6,9],
                [2,4,8],
                [5,7,1]])

print("문제 1")
print("arr의 모양: ",arr.shape)
print("arr의 차원 :",arr.ndim)
print()

print("문제 2")
print("열 방향 평균 : ",arr.mean(axis=0))
print()

print("문제 3")
print("행 방향 최댓값 : ",arr.max(axis=1))
print()

print("문제 4")
print("arr의 전치행렬 : ",arr.T.shape)
print()

print("문제 5")
flat = arr.flatten()
print("arr의 1차원배열 : ",flat.min())
print()

print("문제 6")
print(arr.reshape(1,9))
```
