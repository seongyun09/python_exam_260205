```python
from openpyxl import *

#새로운 엑셀 파일 만들기
w1 = Workbook()

#만들어진 파일 기본 시트 선택
ws = w1.active

w1.save('test.xlsx') #저장

w2 = load_workbook('test.xlsx') # 수정

ws2 = w2.active

#셀에 값 입력하기
#a1셀에 헬로 라는 텍스트입력
ws2["A1"] = 'hello'

#변경사항 저장
w2.save('test.xlsx')

#c3셀에 bye 입력하기
ws2["C3"] = 'bye'
w2.save('test.xlsx')

a = 'python'
#ws2['d5:f8] : d5부터 f8까지의 범위 선택(4행,3열)
for row in ws2['D5':'F8']: #행단위 반복
    for cell in row: #각 행안의셀을 하나씩 꺼내기
        cell.value = a #셀에 파이썬 이라는 값 입력
w2.save('test.xlsx')

num = 1
#A1~A3까지 범위 선택
for row in ws2['A1':'C3']: #행단위 반복
    for cell in row: #각 행안의셀을 하나씩 꺼내기
        cell.value = num
        num += 1 
w2.save('test.xlsx')
```

```python
from openpyxl import *

item1 = {
    '품목' : '바나나',
    '가격' : '3000',
    '수량' : 1,
    '날짜' : '2026-02-06'
}

item2 = {
    '품목' : '파인애플',
    '가격' : '8000',
    '수량' : 5,
    '날짜' : '2026-02-07'
}

item3 = {
    '품목' : '사과',
    '가격' : '5000',
    '수량' : 2,
    '날짜' : '2026-02-08'
}

item4 = {
    '품목' : '레몬',
    '가격' : '5000',
    '수량' : 4,
    '날짜' : '2026-02-09'
}

wb = Workbook()
ws = wb.active

ws['A1'] = '품목'
ws['B1'] = '가격'
ws['C1'] = '수량'
ws['D1'] = '날짜'

items = [item1,item2,item3,item4]
i = 2
for item in items:
    ws['A' + str(i)] = item['품목']
    ws['B' + str(i)] = item['가격']
    ws['C' + str(i)] = item['수량']
    ws['D' + str(i)] = item['날짜']
    i += 1

wb.save('text2.xlsx')
```

```python
from openpyxl import *
from openpyxl.styles import PatternFill

test1 = {
    '이름' : '민수',
    '국어' : 80,
    '영어' : 70,
    '수학' : 90
}

test2 = {
    '이름' : '영희',
    '국어' : 90,
    '영어' : 85,
    '수학' : 88
}

test3 = {
    '이름' : '철수',
    '국어' : 75,
    '영어' : 60,
    '수학' : 70
}

ws['E1'] = '이름'
ws['F1'] = '국어'
ws['G1'] = '영어'
ws['H1'] = '수학'


tests = [test1,test2,test3]
i = 2
for test in tests:
    ws['E' + str(i)] = test['이름']
    ws['F' + str(i)] = test['국어']
    ws['G' + str(i)] = test['영어']
    ws['H' + str(i)] = test['수학']
    i += 1

wb.save('text2.xlsx')
```

```python
from openpyxl import *
from openpyxl.styles import PatternFill

# 셀 배경색 바꾸기 
wb = Workbook()
ws = wb.active

#헤더

ws['E1'] = '이름'
ws['F1'] = '국어'
ws['G1'] = '영어'
ws['H1'] = '수학'

# 노란색 배경
yellow = PatternFill(fill_type='solid',fgColor='FFFF00')

for cell in ws[1]:
    cell.fill = yellow
wb.save('color.xlsx')

#색상코드

# 빨강 : FF0000
# 노랑 : FFFF00
# 파랑 : 0000FF
```
