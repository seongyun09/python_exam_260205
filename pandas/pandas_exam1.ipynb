```python
import pandas as pd

# 1. DataFrame 생성
df = pd.DataFrame({
    "Name": ["Tom", "Jane", "Mike", "Sara", "John", "Emma"],
    "Age": [20, 22, 19, 21, 23, 20],
    "Score": [85, 92, 78, 88, 95, 70],
    "Grade": ["B", "A", "C", "B", "A", "C"],
})

# 2. 기본 정보 확인
print("--- 2. 기본 정보 확인 ---")
print(df)               # 전체 출력
print(df.shape)         # 모양 (행, 열)
df.info()               # 정보 (결측치, 데이터 타입 등)
print(df.describe())    # 기술 통계

# 3. 열 선택
print("\n--- 3. 열 선택 ---")
print(df["Score"])             # Score 열만
print(df[["Name", "Age"]])     # Name, Age 열만 (리스트 활용)

# 4. 행 선택
print("\n--- 4. 행 선택 ---")
print(df.iloc[0])              # 첫 번째 행
print(df.iloc[-1])             # 마지막 행

# index 변경
df.index = ["학생1", "학생2", "학생3", "학생4", "학생5", "학생6"]
print(df.loc["학생3"])         # loc를 사용한 특정 인덱스 행 선택

# 5. 통계 계산
print("\n--- 5. 통계 계산 ---")
print(f"최대 점수: {df['Score'].max()}")
print(f"최소 점수: {df['Score'].min()}")
print(f"평균 점수: {df['Score'].mean()}")

# 6. 조건 선택
print("\n--- 6. 조건 선택 ---")
print(df[df["Score"] >= 85])                 # 85점 이상
print(df[df["Grade"] == "A"])                # Grade가 A
print(df[df["Score"] > df["Score"].mean()])  # 평균 초과 학생

# 7. 일부만 보기
print("\n--- 7. 일부만 보기 ---")
print(df.head(3))  # 상위 3개
print(df.tail(2))  # 하위 2개
```
