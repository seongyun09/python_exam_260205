```python
import numpy as np

arr = np.array([[1.5, 2.5, 3.5, 4.5],
                [5.5, 6.5, 7.5, 8.5]])

# 1
print(arr.dtype)
print(arr.itemsize)

# 2
print(arr.sum(axis=0))

# 3
print(arr.mean(axis=1))

# 4
reshaped = arr.reshape(4, 2)
print(reshaped)

# 5
print(reshaped.T.shape)

# 6
print(arr.flatten().shape)
```
