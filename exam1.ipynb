```python
class Employee:
    def work(self):
        print("직원이 일을 합니다")
        # 코드를 작성해주세요.

class Manager(Employee):
    def work(self):
        print("메니저가 팀을 관리 합니다")
        # 코드를 작성해주세요.

manager = Manager()
manager.work()

# 출력 결과:
# 직원이 일을 합니다.
# 매니저가 팀을 관리합니다.
```
